```python
import torch
import matplotlib.pyplot as plt

from fim.models.fim_model import FIMModel
from fim.utils.seed import set_seed

set_seed(42)
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
model = FIMModel(
    in_channels=1,
    out_channels=1,
    latent_channels=16,
    trace_dim=8,
    hidden_channels=32,
    memory_threshold=0.2,
).to(device)

model.eval()
```

```python
steps = 50
x = torch.randn(1, 1, 16, 16).to(device)

states = []
norms = []

with torch.no_grad():
    for _ in range(steps):
        y, out = model.step(x, update_memory=True)
        states.append(out.state.cpu())
        norms.append(torch.norm(out.state).item())
        x = y

plt.plot(norms)
plt.title("Fabric Norm Over Time")
plt.show()
```

```python
print("Final norm:", norms[-1])

if norms[-1] > 1e4:
    print("⚠️ Potential instability detected")
else:
    print("✅ Stable dynamics")
```
